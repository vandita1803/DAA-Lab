```python
import random
import time
import matplotlib.pyplot as plt

def majority_element(arr):
    candidate = None
    count = 0

    for num in arr:
        if count == 0:
            candidate = num
            count = 1
        elif num == candidate:
            count += 1
        else:
            count -= 1

    count = 0
    for num in arr:
        if num == candidate:
            count += 1

    if count > len(arr) // 2:
        return candidate
    else:
        return None


sizes = list(range(100, 10001, 500))

avg_times = []

runs = 20

for size in sizes:

    total_time = 0

    for _ in range(runs):


        majority = random.randint(1, 10)


        majority_count = size // 2 + 1

        arr = [majority] * majority_count


        while len(arr) < size:
            value = random.randint(1, 10)
            arr.append(value)

        random.shuffle(arr)

        start = time.perf_counter_ns()

        majority_element(arr)

        end = time.perf_counter_ns()

        total_time += (end - start)

    avg_times.append(total_time / runs)
```

```python
plt.figure(figsize=(10,6))

plt.plot(
    sizes,
    avg_times,
    marker='o',
    linewidth=2,
    markersize=5
)

plt.title("Moore's Voting Algorithm")
plt.xlabel("Input Size (n)")
plt.ylabel("Average Execution Time (ns)")
plt.grid(True)

plt.show()
```
