```python
import random
import time
import matplotlib.pyplot as plt

def dutch_national_flag(arr):
    low = 0
    mid = 0
    high = len(arr) - 1

    while mid <= high:

        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1

        elif arr[mid] == 1:
            mid += 1

        else:
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1



sizes = list(range(100, 10001, 500))

avg_times = []

runs = 20

for size in sizes:

    total_time = 0

    for _ in range(runs):


        arr = [random.randint(0,2) for _ in range(size)]

        start = time.perf_counter_ns()

        dutch_national_flag(arr)

        end = time.perf_counter_ns()

        total_time += (end - start)

    avg_times.append(total_time / runs)
    print(arr)
```

```python
plt.figure(figsize=(10,6))

plt.plot(
    sizes,
    avg_times,
    marker='o',
    linewidth=2,
    markersize=5
)

plt.title("Dutch National Flag Algorithm")
plt.xlabel("Input Size (n)")
plt.ylabel("Average Execution Time (ns)")
plt.grid(True)

plt.show()
```
