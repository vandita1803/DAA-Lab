```python
import time
import statistics


def partition(arr, low, high):

    pivot = arr[high]
    i = low

    for j in range(low, high):

        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[high] = arr[high], arr[i]

    return i


def quick_select(arr, k):

    # Convert kth largest to index
    target = len(arr) - k

    low = 0
    high = len(arr) - 1

    while low <= high:

        pivot_index = partition(arr, low, high)

        if pivot_index == target:
            return arr[pivot_index]

        elif pivot_index < target:
            low = pivot_index + 1

        else:
            high = pivot_index - 1

    return None


# -----------------------------------
# Performance Analysis
# -----------------------------------

sizes = list(range(100, 10001, 500))

times = []

runs = 5

for n in sizes:

    # Fixed deterministic array
    arr = list(range(n, 0, -1))

    # Find the middle-largest element
    k = n // 2

    measurements = []

    for _ in range(runs):

        # Copy so every run receives
        # the same original input
        test_arr = arr.copy()

        start = time.perf_counter_ns()

        quick_select(test_arr, k)

        end = time.perf_counter_ns()

        measurements.append(end - start)

    times.append(statistics.median(measurements))

    print(f"n = {n} completed")
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))

plt.plot(
    sizes,
    times,
    marker='o'
)

plt.title("Kth Largest using Quick Select")
plt.xlabel("Input Size (n)")
plt.ylabel("Median Execution Time (ns)")
plt.grid(True)

plt.show()
```
