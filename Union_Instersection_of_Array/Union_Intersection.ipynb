```python
import random
import time
import matplotlib.pyplot as plt

# Union and Intersection
def union_intersection(A, B):

    i = 0
    j = 0

    union = []
    intersection = []

    while i < len(A) and j < len(B):

        if A[i] < B[j]:
            union.append(A[i])
            i += 1

        elif A[i] > B[j]:
            union.append(B[j])
            j += 1

        else:
            union.append(A[i])
            intersection.append(A[i])
            i += 1
            j += 1

    # Remaining elements of A
    while i < len(A):
        union.append(A[i])
        i += 1

    # Remaining elements of B
    while j < len(B):
        union.append(B[j])
        j += 1

    return union, intersection


# Input size
sizes = list(range(100, 5001, 250))

avg_times = []

runs = 20

for size in sizes:

    A = sorted(random.randint(1, 100000) for _ in range(size))
    B = sorted(random.randint(1, 100000) for _ in range(size))

    total_time = 0

    for _ in range(runs):

        start = time.perf_counter_ns()

        union_intersection(A, B)

        end = time.perf_counter_ns()

        total_time += end - start

    avg_times.append(total_time / runs)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    sizes,
    avg_times,
    marker='o',
    linewidth=2,
    markersize=5
)

plt.title("Union and Intersection of Two Sorted Arrays")
plt.xlabel("Input Size (n)")
plt.ylabel("Average Execution Time (ns)")
plt.grid(True)

plt.show()
```
