```python
import time
import statistics


def max_product_subarray(arr):

    max_product = arr[0]
    min_product = arr[0]
    result = arr[0]

    for i in range(1, len(arr)):

        if arr[i] < 0:
            max_product, min_product = min_product, max_product

        max_product = max(arr[i], max_product * arr[i])
        min_product = min(arr[i], min_product * arr[i])

        result = max(result, max_product)

    return result


# -----------------------------------
# Performance Analysis
# -----------------------------------

sizes = list(range(100, 10001, 500))

times = []

runs = 5

# Fixed repeating pattern containing
# positive, negative and zero values
pattern = [-2, 3, -1, 4, 0, 2, -3, 1]

for n in sizes:

    # Same deterministic input for every
    # measurement of this input size
    arr = (pattern * ((n // len(pattern)) + 1))[:n]

    measurements = []

    for _ in range(runs):

        start = time.perf_counter_ns()

        max_product_subarray(arr)

        end = time.perf_counter_ns()

        measurements.append(end - start)

    times.append(statistics.median(measurements))

    print(f"n = {n} completed")
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))

plt.plot(
    sizes,
    times,
    marker='o'
)

plt.title("Maximum Product Subarray")
plt.xlabel("Input Size (n)")
plt.ylabel("Median Execution Time (ns)")
plt.grid(True)

plt.show()
```
