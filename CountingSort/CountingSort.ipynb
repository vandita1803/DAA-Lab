```python
import time
import statistics

def counting_sort(arr):
    if not arr:
        return []

    max_value = max(arr)
    count = [0] * (max_value + 1)

    for value in arr:
        count[value] += 1

    result = []

    for value in range(len(count)):
        for _ in range(count[value]):
            result.append(value)

    return result


sizes = list(range(100, 10001, 500))
times = []
runs = 5

for n in sizes:
    # Fixed deterministic input
    arr = [(i * 37) % n for i in range(n)]

    measurements = []

    for _ in range(runs):
        start = time.perf_counter_ns()
        counting_sort(arr)
        end = time.perf_counter_ns()

        measurements.append(end - start)

    times.append(statistics.median(measurements))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(sizes, times, marker='o')
plt.title("Counting Sort")
plt.xlabel("Input Size (n)")
plt.ylabel("Median Execution Time (ns)")
plt.grid(True)
plt.show()
```
