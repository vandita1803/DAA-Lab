```python
import time
import statistics

def can_allocate(books, students, max_pages):

    student_count = 1
    current_pages = 0

    for pages in books:

        if current_pages + pages <= max_pages:
            current_pages += pages

        else:
            student_count += 1
            current_pages = pages

            if student_count > students:
                return False

    return True


# Allocate Minimum Pages
def allocate_pages(books, students):

    low = max(books)
    high = sum(books)

    answer = high

    while low <= high:

        mid = (low + high) // 2

        if can_allocate(books, students, mid):
            answer = mid
            high = mid - 1

        else:
            low = mid + 1

    return answer


sizes = list(range(100, 5001, 250))

times = []

runs = 5

for n in sizes:

    books = [100 + (i * 37) % 901 for i in range(n)]

    students = max(2, n // 10)

    measurements = []

    for _ in range(runs):

        start = time.perf_counter_ns()

        allocate_pages(books, students)

        end = time.perf_counter_ns()

        measurements.append(end - start)

    times.append(statistics.median(measurements))

    print(f"n = {n} completed")
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))

plt.plot(
    sizes,
    times,
    marker='o'
)

plt.title("Allocate Minimum Pages")
plt.xlabel("Number of Books (n)")
plt.ylabel("Median Execution Time (ns)")
plt.grid(True)

plt.show()
```
