```python
import time
import statistics

def smallest_missing_positive(arr):
    n = len(arr)

    for i in range(n):
        while 1 <= arr[i] <= n and arr[arr[i] - 1] != arr[i]:
            correct_index = arr[i] - 1
            arr[i], arr[correct_index] = arr[correct_index], arr[i]

    for i in range(n):
        if arr[i] != i + 1:
            return i + 1

    return n + 1


sizes = list(range(100, 10001, 500))
times = []
runs = 5

for n in sizes:
    # Fixed deterministic input
    arr = list(range(2, n + 2))
    arr[n // 2] = 1

    measurements = []

    for _ in range(runs):
        test_arr = arr.copy()

        start = time.perf_counter_ns()
        smallest_missing_positive(test_arr)
        end = time.perf_counter_ns()

        measurements.append(end - start)

    times.append(statistics.median(measurements))
    print(f"n = {n} completed")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(sizes, times, marker='o')
plt.title("Smallest Missing Positive")
plt.xlabel("Input Size (n)")
plt.ylabel("Median Execution Time (ns)")
plt.grid(True)
plt.show()
```
